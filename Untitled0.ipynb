```python
import csv
import tempfile
import zipfile
from pathlib import Path
SOURCE = "upload"  # "upload": 내 컴퓨터에서 ZIP 선택 / "drive": Google Drive 사용
DRIVE_PATH = "/content/drive/MyDrive/pokemon_data.zip"

def prepare_data(source, extract_parent):
    # 매번 새 폴더에 풀어 예전 파일과 섞이지 않게 하고, 폴더째 압축해도 찾습니다.
    if source.is_file() and zipfile.is_zipfile(source):
        search_root = Path(tempfile.mkdtemp(prefix="pokemon_", dir=extract_parent))
        with zipfile.ZipFile(source) as archive:
            for entry in archive.infolist():
                target = (search_root / entry.filename.replace("\\", "/")).resolve()
                if not target.is_relative_to(search_root.resolve()):
                    raise ValueError("ZIP에 대상 폴더 밖으로 나가는 경로가 있습니다.")
            archive.extractall(search_root)
    elif source.is_dir():
        search_root = source
    else:
        raise FileNotFoundError(f"ZIP 또는 데이터 폴더를 확인하세요: {source}")
    candidates = [p.parent for p in search_root.rglob("train.csv")
                  if (p.parent / "test.csv").is_file()
                  and (p.parent / "train").is_dir() and (p.parent / "test").is_dir()]
    if len(candidates) != 1:
        raise ValueError(f"데이터 폴더를 {len(candidates)}개 찾았습니다. train/test 폴더와 CSV가 한곳에 있는지 확인하세요.")
    return candidates[0]

if SOURCE == "upload":
    uploaded = files.upload()  # 사용자가 고른 ZIP을 코랩 런타임에 업로드합니다.
    paths = [Path(name).resolve() for name in uploaded]
    del uploaded  # 사진 파일 바이트를 메모리에 계속 보관하지 않습니다.
    if len(paths) != 1 or paths[0].suffix.lower() != ".zip":
        raise ValueError("사진 폴더와 CSV가 함께 들어 있는 ZIP 파일 하나를 선택하세요.")
    source_path = paths[0]
elif SOURCE == "drive":
    from google.colab import drive
    drive.mount("/content/drive")
    source_path = Path(DRIVE_PATH)
else:
    raise ValueError('SOURCE는 "upload" 또는 "drive"로 지정하세요.')
DATA_DIR = prepare_data(source_path, Path("/content"))
print("데이터 위치:", DATA_DIR)
```

```python
def check_data(root):
    # 학습 중간에 파일 누락으로 멈추지 않도록 열 이름·중복 ID·사진 경로를 먼저 검사합니다.
    for split in ("train", "test"):
        with (root / f"{split}.csv").open(encoding="utf-8-sig", newline="") as f:
            reader = csv.DictReader(f)
            required = {"id", "label"} if split == "train" else {"id"}
            if not required.issubset(reader.fieldnames or []):
                raise ValueError(f"{split}.csv에는 {sorted(required)} 열이 필요합니다.")
            records = list(reader)
        if not records or any(not row.get(key) for row in records for key in required):
            raise ValueError(f"{split}.csv에 비어 있는 ID 또는 정답이 있습니다.")
        ids = [row["id"] for row in records]
        if len(ids) != len(set(ids)) or any("/" in i or "\\" in i or i in {".", ".."} for i in ids):
            raise ValueError(f"{split}.csv의 ID가 중복되거나 파일명 형식이 아닙니다.")
        missing = [i for i in ids if not (root / split / f"{i}.jpg").is_file()]
        if missing:
            raise FileNotFoundError(f"{split}: 사진 {len(missing)}장이 없습니다. 예: {missing[:3]}")
        print(f"{split}: CSV {len(records):,}행, 사진 연결 정상")
check_data(DATA_DIR)
```

```python
import csv
import random
from pathlib import Path
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
# DATA_DIR은 위 업로드 셀에서 찾은 폴더를 그대로 사용합니다.
OUT = Path("/content/pokemon_results")  # 재실험을 보관하려면 폴더 이름을 바꾸세요.
OUT.mkdir(parents=True, exist_ok=True)
SIZE, BATCH, EPOCHS = 128, 32, 100
torch.manual_seed(42)  # 초기 가중치·셔플의 난수 시작점을 고정합니다.
torch.set_num_threads(4)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
def read_csv(path):
    # DictReader로 열 순서 대신 열 이름을 쓰고, BOM이 있는 CSV도 읽습니다.
    with path.open(encoding="utf-8-sig", newline="") as f:
        return list(csv.DictReader(f))
rows, test_rows = read_csv(DATA_DIR / "train.csv"), read_csv(DATA_DIR / "test.csv")
classes = sorted({row["label"] for row in rows})  # 번호 순서를 일정하게 만듭니다.
label_to_id = {name: i for i, name in enumerate(classes)}
```

```python
def split_rows(rows, fraction=0.2, seed=42):
    # 종류별로 나눠 모든 포켓몬을 양쪽에 포함하고, 원본 목록은 바꾸지 않습니다.
    rng = random.Random(seed)
    train, val = [], []
    for name in sorted({row["label"] for row in rows}):
        group = [row for row in rows if row["label"] == name]
        if len(group) < 2:
            raise ValueError(f"{name}: 분리를 위해 사진이 2장 이상 필요합니다.")
        rng.shuffle(group)
        count = min(len(group) - 1, max(1, round(len(group) * fraction)))
        train.extend(group[count:])
        val.extend(group[:count])
    return train, val
train_rows, val_rows = split_rows(rows)
print(f"{device} | {len(classes)}종 | 학습 {len(train_rows)}장, 검증 {len(val_rows)}장")
```

```python
def make_transform(training=False):
    # 학습에만 증강하여 과적합을 줄이고, 평가는 매번 동일한 입력으로 합니다.
    steps = [transforms.Resize((SIZE, SIZE))]
    if training:
        steps.append(transforms.RandomHorizontalFlip())
    return transforms.Compose(steps + [transforms.ToTensor()])
class PokemonDataset(Dataset):
    def __init__(self, rows, folder, transform):
        # 전부 읽지 않고 경로와 규칙만 보관하여 메모리를 아낍니다.
        self.rows, self.folder, self.transform = rows, folder, transform
    def __len__(self):
        return len(self.rows)  # DataLoader가 전체 샘플 수를 알 수 있게 합니다.
    def __getitem__(self, index):
        # RGB로 채널 수를 3개로 통일하고, 파일은 읽은 뒤 자동으로 닫습니다.
        row = self.rows[index]
        with Image.open(self.folder / (row["id"] + ".jpg")) as image:
            tensor = self.transform(image.convert("RGB"))
        target = label_to_id[row["label"]] if "label" in row else row["id"]
        return tensor, target
train_ds = PokemonDataset(train_rows, DATA_DIR / "train", make_transform(True))
val_ds = PokemonDataset(val_rows, DATA_DIR / "train", make_transform())
train_loader = DataLoader(train_ds, batch_size=BATCH, shuffle=True, num_workers=0)
val_loader = DataLoader(val_ds, batch_size=BATCH, shuffle=False, num_workers=0)
```

```python
def make_cnn(num_classes):
    # 같은 구조는 반복문으로 만들고, 이전 출력 채널을 다음 입력으로 연결합니다.
    layers, channels = [], 3
    for width in (32, 64, 128, 256):
        layers += [nn.Conv2d(channels, width, 3, padding=1),
                   nn.BatchNorm2d(width), nn.ReLU(), nn.MaxPool2d(2)]
        channels = width
    # 평균 풀링으로 큰 특징 지도를 256개 값으로 줄여 분류층을 작게 만듭니다.
    layers += [nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Dropout(0.3),
               nn.Linear(256, num_classes)]
    return nn.Sequential(*layers)
model = make_cnn(len(classes)).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=0.0001, weight_decay=1e-6)
```

```python
def run_epoch(model, loader, optimizer=None):
    # 옵티마이저 유무로 학습/검증을 구분해 같은 반복 코드를 재사용합니다.
    training = optimizer is not None
    model.train(training)
    loss_sum, correct, total = 0.0, 0, 0
    with torch.set_grad_enabled(training):  # 검증에는 역전파 계산이 필요 없습니다.
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            if training:
                optimizer.zero_grad(set_to_none=True)  # 이전 배치의 기울기를 지웁니다.
            logits = model(images)
            loss = criterion(logits, labels)
            if training:
                loss.backward()
                optimizer.step()
            total += labels.size(0)
            loss_sum += loss.item() * labels.size(0)  # 배치 크기가 달라도 정확히 평균냅니다.
            correct += (logits.argmax(1) == labels).sum().item()
    return loss_sum / total, correct / total
```

```python
best, history = -1.0, []
for epoch in range(1, EPOCHS + 1):
    train_loss, train_acc = run_epoch(model, train_loader, optimizer)
    val_loss, val_acc = run_epoch(model, val_loader)
    history.append([epoch, train_loss, train_acc, val_loss, val_acc])
    print(f"{epoch:02d}: train={train_acc:.2%}, val={val_acc:.2%}, val_loss={val_loss:.4f}", flush=True)
    if val_acc > best:  # 마지막 모델보다 검증에서 가장 잘한 모델을 남깁니다.
        best = val_acc
        torch.save({"weights": model.state_dict(), "classes": classes, "size": SIZE}, OUT / "best.pt")
with (OUT / "history.csv").open("w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["epoch", "train_loss", "train_acc", "val_loss", "val_acc"])
    writer.writerows(history)
```

```python
checkpoint = torch.load(OUT / "best.pt", map_location="cpu", weights_only=True)
classes, SIZE = checkpoint["classes"], checkpoint["size"]
model = make_cnn(len(classes)).to(device)
model.load_state_dict(checkpoint["weights"])
model.eval()  # 예측 때는 Dropout을 끄고 BatchNorm의 저장 통계를 사용합니다.
test_ds = PokemonDataset(test_rows, DATA_DIR / "test", make_transform())
test_loader = DataLoader(test_ds, batch_size=BATCH, shuffle=False, num_workers=0)
results = []
with torch.inference_mode():
    for images, ids in test_loader:
        indexes = model(images.to(device)).argmax(1).cpu().tolist()
        results.extend((image_id, classes[i]) for image_id, i in zip(ids, indexes))
assert [row[0] for row in results] == [row["id"] for row in test_rows]
with (OUT / "submission.csv").open("w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["id", "label"])
    writer.writerows(results)
print(f"저장 완료: {OUT / 'submission.csv'} ({len(results)}장)")
```
